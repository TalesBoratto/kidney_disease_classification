# Ensure Annotations

```python
# Ensure Annotation
def get_predict(x:int, y:int) -> int:
    return x*y
```

```python
get_predict(x=2, y=3)
```

```python
get_predict(x=2, y="3")
```

```python
from ensure import ensure_annotations

@ensure_annotations
def get_predict(x:int, y:int) -> int:
    return x*y
```

```python
get_predict(x=2, y=3)
```

```python
get_predict(x=2, y="3")
```
